```python
# torch device
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
# make sure that there will be only one StandardScaler   --- Singleton Class. 

class GlobalScaler:
    _instance = None # class variable

    def __init__(self): # constructor
        if GlobalScaler._instance is not None:
            raise Exception ('GlobalScaler is Singleton Class')
        
        self.scaler = StandardScaler()

    # singleton pattern --- not instantiating using GlobalScaler()
    @classmethod
    def get_instance(cls):
        if cls._instance is None:
            cls._instance = GlobalScaler()
        return cls._instance
```

```python
# class to create dataset 
class FashionDS(Dataset):
    global_scaler = GlobalScaler.get_instance() # protected variable _instance, so use getter

    def __init__(self, dataframe, device = device, 
                        is_train = True, label_col = None):
        self.df = dataframe
        self.device = device
        self.is_train = is_train
        self.scaler = self.global_scaler.scaler   
        self.label_col = label_col
    
        self.labels = self.df[label_col].to_numpy()  # dataframe to numpy
        
        if self.is_train:                              # for training data
            X = self.df.drop(label_col, axis=1)            
            self.features = self.scaler.fit_transform(X)
        else:           
            X = self.df.drop(label_col, axis=1)          # for testing data
            self.features = self.scaler.transform(X)    
        
    def __len__(self): # len of dataframe
        return len(self.features)
    
    def __getitem__(self, index): # return torch tensor
        feature = self.features[index]
        label = self.labels[index]

        feature = torch.tensor(feature, dtype=torch.float32, device=self.device)
        label = torch.tensor(label, dtype=torch.int64, device=device)
        return feature, label
    
    
```

```python
# resetting dataframe index
train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)
```

```python
# converting to dataset
label_col = 'label'
train_ds = FashionDS(train_df, is_train=True, label_col=label_col)
test_ds = FashionDS(test_df, is_train=False, label_col=label_col)
```

```python
type(train_df), type(train_ds), type(test_df), type(test_ds) # checking data types
```

```python
# creating batches using dataloader
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=True)
```

```python
# defining model
class Model(nn.Module):

    def __init__(self, input_dim): # constructor
        super().__init__()

        # dropout rates
        dor1 = 0.1
        dor2 = 0.2
        dor3 = 0.3
        dor4 = 0.4


        # first layer  -- gets input
        out_dim_1 = 584
        self.layer1 = nn.Linear(input_dim, out_dim_1)   # output 584
        self.bn1 = nn.BatchNorm1d(out_dim_1)    # batchnorm
        self.actv1 = nn.ReLU() # activation
        self.do1 = nn.Dropout(dor1)  # dropout

        # second layer
        out_dim_2 = 384
        self.layer2 = nn.Linear(out_dim_1, out_dim_2)   # output 384
        self.bn2 = nn.BatchNorm1d(out_dim_2)
        self.actv2 = nn.ReLU()
        self.do2 = nn.Dropout(dor2)

        # third layer
        out_dim_3 = 192
        self.layer3 = nn.Linear(out_dim_2, out_dim_3)   # output 192
        self.bn3 = nn.BatchNorm1d(out_dim_3)
        self.actv3 = nn.ReLU()
        self.do3 = nn.Dropout(dor3)

        # fourth layer
        out_dim_4 = 90
        self.layer4 = nn.Linear(out_dim_3, out_dim_4)   # output 90
        self.bn4 = nn.BatchNorm1d(out_dim_4)
        self.actv4 = nn.ReLU()
        self.do4 = nn.Dropout(dor4)

        # fifth layer -- gives outputs
        self.layer5 = nn.Linear(out_dim_4, 10)           # output 10

    
    # forward propagation
    def forward(self, x):

        x = self.layer1(x)  # aggregation
        x = self.bn1(x)  # batch normalization
        x = self.actv1(x)   # activation
        x = self.do1(x)   # dropout

        x = self.layer2(x)
        x = self.bn2(x)
        x = self.actv2(x)
        x = self.do2(x)

        x = self.layer3(x)
        x = self.bn3(x)
        x = self.actv3(x)
        x = self.do3(x)

        x = self.layer4(x)
        x = self.bn4(x)
        x = self.actv4(x)
        x = self.do4(x)


        x = self.layer5(x)
        return x
    

model = Model(input_dim=784).to(device=device) # sending model to device
print(model)
        
```

```python
loss_fn = nn.CrossEntropyLoss() # loss function
optimizer = torch.optim.Adam(model.parameters(), lr=ALPHA, weight_decay=1e-5) # optimizer

# learning rate scheduler
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer=optimizer,
                                                       min_lr=MIN_LR,
                                                       mode='min', factor=LR_FACTOR, 
                                                       patience=PATIENCE)

# to store values
loss, test_loss = [], []
acc, test_acc = [], []
n_epoch = []

for epoch in range(EPOCHS): # model training over epochs
    model.train() # set to training mode
    # initialize loss and accuracy for a batch to 0
    train_loss = 0.0
    train_acc = 0.0

    for i, data in enumerate(train_loader): # loop over batches
        inputs, labels = data
        optimizer.zero_grad() # make gradients of previous steps 0
        outputs = model(inputs) # model output
        preds = torch.argmax(outputs, dim=1) # prediction

        batch_loss = loss_fn(outputs, labels) # loss for batch
        batch_acc = accuracy_score(labels.cpu().numpy(), preds.cpu().numpy()) # accuracy for batch

        batch_loss.backward() # backpropagation of loss
        optimizer.step() # update weights and biases
        
        train_loss += batch_loss.item() * inputs.size(0)
        train_acc += batch_acc * inputs.size(0)

    
    train_loss = train_loss/len(train_ds)
    train_acc = train_acc/len(train_ds)
    # store average batch loss and accuracy
    loss.append(train_loss)
    acc.append(train_acc)


    # testing now
    with torch.no_grad(): # stop computing gradients
        model.eval() # set model to evaluation mode
        tst_loss = 0.0
        tst_acc = 0.0

        for i, data in enumerate(test_loader): # loop over batches
            inputs, labels = data
            outputs = model(inputs) # model outputs
            preds = torch.argmax(outputs, dim=1) # predictions

            batch_loss = loss_fn(outputs, labels) # batch loss
            batch_acc = accuracy_score(labels.cpu().numpy(), preds.cpu().numpy()) # batch accuracy

            tst_loss += batch_loss.item() * inputs.size(0)
            tst_acc += batch_acc * inputs.size(0)

        tst_loss = tst_loss/len(test_ds)
        tst_acc = tst_acc/len(test_ds)

        test_loss.append(tst_loss)
        test_acc.append(tst_acc)

    n_epoch.append(epoch)

    scheduler.step(tst_loss) # update learning rate

    # print status
    if epoch%10 == 0: 
        print(f'Epoch {epoch}: loss = {train_loss:.4f}, accuracy = {train_acc:.3f}'+
                        f', val_loss = {tst_loss:.4f}, val_accuracy = {tst_acc:.3f}')
        
        
```

```python
# saving model
torch.save(model.state_dict(), 'torch_mnist.pt')
```

```python
y_true, y_pred = [], []
# prediction on test data
for feature, label in test_loader:
    pred = model(feature).detach().cpu().numpy()
    y_true.extend(label)
    y_pred.extend(pred.argmax(axis=1))

print(classification_report(y_true, y_pred))
```
